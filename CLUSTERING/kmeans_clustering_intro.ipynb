El **Clustering K-Means** agrupa puntos de datos similares en clústeres sin necesidad de datos etiquetados. Se utiliza para descubrir patrones ocultos cuando el objetivo es organizar datos en función de la similitud.

- Ayuda a identificar agrupaciones naturales en conjuntos de datos no etiquetados.
- Funciona agrupando puntos en función de la distancia a los centros de los clústeres.
- Comúnmente utilizado en segmentación de clientes, compresión de imágenes y descubrimiento de patrones.
- Útil cuando se necesita estructura a partir de datos crudos y desorganizados.

## Funcionamiento del Clustering K-Means

Supongamos que tenemos un conjunto de datos de elementos con ciertas características y valores para dichas características, como un vector. La tarea consiste en categorizar esos elementos en grupos. Para lograrlo, usaremos el algoritmo K-Means. **"k"** representa el número de grupos o clústeres en los que queremos clasificar nuestros elementos.

El algoritmo categorizará los elementos en **"k" grupos o clústeres de similitud**. Para calcular esa similitud, usaremos la **distancia euclidiana** como medida. El algoritmo funciona de la siguiente manera:

1. **Inicialización:** Comenzamos seleccionando aleatoriamente *k* centroides de clúster.
2. **Paso de asignación:** Cada punto de datos se asigna al centroide más cercano, formando clústeres.
3. **Paso de actualización:** Tras la asignación, recalculamos el centroide de cada clúster promediando los puntos que contiene.
4. **Repetición:** Este proceso se repite hasta que los centroides ya no cambian o se alcanza el número máximo de iteraciones.

El objetivo es particionar el conjunto de datos en *k* clústeres de forma que los puntos de datos dentro de cada clúster sean más similares entre sí que a los de otros clústeres.

> **Nota:** Seleccionar el número correcto de clústeres es importante para una segmentación significativa. Para ello se usa el **Método del Codo** (*Elbow Method*), una herramienta gráfica que determina el valor óptimo de *k* en K-Means.

## ¿Por Qué Usar el Clustering K-Means?

K-Means es popular en una gran variedad de aplicaciones debido a su simplicidad, eficiencia y efectividad. Aquí se explica por qué se usa ampliamente:

1. **Segmentación de datos:** Uno de los usos más comunes de K-Means es segmentar los datos en grupos distintos. Por ejemplo, las empresas usan K-Means para agrupar clientes según su comportamiento, como patrones de compra o interacción con el sitio web.

2. **Compresión de imágenes:** K-Means puede usarse para reducir la complejidad de las imágenes agrupando píxeles similares en clústeres, comprimiendo eficazmente la imagen. Esto es útil para el almacenamiento y procesamiento de imágenes.

3. **Detección de anomalías:** K-Means puede aplicarse para detectar anomalías o valores atípicos (*outliers*) identificando puntos de datos que no pertenecen a ninguno de los clústeres.

4. **Agrupación de documentos:** En el procesamiento del lenguaje natural (PLN), K-Means se usa para agrupar documentos o artículos similares. Se usa frecuentemente en aplicaciones como sistemas de recomendación o categorización de noticias.

5. **Organización de grandes conjuntos de datos:** Al trabajar con grandes conjuntos de datos, K-Means puede ayudar a organizar los datos en fragmentos más pequeños y manejables basándose en similitudes, mejorando la eficiencia del análisis de datos.

## Implementación del Clustering K-Means

Usaremos el conjunto de datos *blobs* y mostraremos cómo se forman los clústeres usando el lenguaje de programación Python.

---

### Paso 1: Importar las librerías necesarias

Importaremos las siguientes librerías:

- **NumPy:** para operaciones numéricas (por ejemplo, cálculo de distancias).
- **Matplotlib:** para graficar los datos y los resultados.
- **Scikit-learn:** para crear un conjunto de datos sintético usando `make_blobs`.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

### Paso 2: Crear un Conjunto de Datos Personalizado

Generaremos un conjunto de datos sintético con `make_blobs`.

- **`make_blobs(n_samples=500, n_features=2, centers=3)`:** Genera 500 puntos de datos en un espacio 2D, agrupados en 3 clústeres.
- **`plt.scatter(X[:, 0], X[:, 1])`:** Grafica el conjunto de datos en 2D, mostrando todos los puntos.
- **`plt.show()`:** Muestra el gráfico.

```python
X, y = make_blobs(n_samples=500, n_features=2, centers=3, random_state=23)

fig = plt.figure(0)
plt.grid(True)
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

**Salida:** Se muestra el conjunto de datos de clustering con los 500 puntos distribuidos en 3 grupos naturales en el espacio 2D.

---

### Paso 3: Inicializar Centroides Aleatorios

Inicializaremos aleatoriamente los centroides para el clustering K-Means.

- **`np.random.seed(23)`:** Garantiza la reproducibilidad fijando la semilla aleatoria.
- El bucle `for` inicializa *k* centroides aleatorios con valores entre -2 y 2, para un conjunto de datos 2D.

```python
k = 3

clusters = {}
np.random.seed(23)

for idx in range(k):
    center = 2 * (2 * np.random.random((X.shape[1],)) - 1)
    points = []
    cluster = {
        'center': center,
        'points': []
    }
    clusters[idx] = cluster

clusters
```

**Salida:** Se muestra un diccionario con los 3 clústeres iniciales, cada uno con un centro aleatorio y una lista de puntos vacía.

---

### Paso 4: Graficar los Centroides Inicializados Aleatoriamente con los Puntos de Datos

Ahora graficamos los puntos de datos junto con los centroides iniciales.

- **`plt.grid()`:** Dibuja una cuadrícula.
- **`plt.scatter(center[0], center[1], marker='*', c='red')`:** Grafica el centro del clúster como una estrella roja (marcador `*`).

```python
plt.scatter(X[:, 0], X[:, 1])
plt.grid(True)
for i in clusters:
    center = clusters[i]['center']
    plt.scatter(center[0], center[1], marker='*', c='red')
plt.show()
```

**Salida:** Se muestran los puntos de datos junto con los 3 centroides iniciales aleatorios marcados como estrellas rojas.

---

### Paso 5: Definir la Distancia Euclidiana

Para asignar puntos de datos al centroide más cercano, definimos una función de distancia:

- **`np.sqrt()`:** Calcula la raíz cuadrada de un número o de un array elemento a elemento.
- **`np.sum()`:** Suma todos los elementos de un array o a lo largo de un eje especificado.

```python
def distance(p1, p2):
    return np.sqrt(np.sum((p1 - p2) ** 2))
```

### Paso 6: Crear las Funciones de Asignación y Actualización

A continuación, definimos funciones para asignar puntos al centroide más cercano y actualizar los centroides basándose en el promedio de los puntos asignados a cada clúster.

- **`dist.append(dis)`:** Añade la distancia calculada a la lista `dist`.
- **`curr_cluster = np.argmin(dist)`:** Encuentra el índice del clúster más cercano seleccionando la distancia mínima.
- **`new_center = points.mean(axis=0)`:** Calcula el nuevo centroide tomando la media de los puntos del clúster.

```python
def assign_clusters(X, clusters):
    for idx in range(X.shape[0]):
        dist = []

        curr_x = X[idx]

        for i in range(k):
            dis = distance(curr_x, clusters[i]['center'])
            dist.append(dis)
        curr_cluster = np.argmin(dist)
        clusters[curr_cluster]['points'].append(curr_x)
    return clusters


def update_clusters(X, clusters):
    for i in range(k):
        points = np.array(clusters[i]['points'])
        if points.shape[0] > 0:
            new_center = points.mean(axis=0)
            clusters[i]['center'] = new_center

            clusters[i]['points'] = []
    return clusters
```

### Paso 7: Predecir el Clúster para los Puntos de Datos

Creamos una función para predecir el clúster de cada punto de datos basándose en los centroides finales.

- **`pred.append(np.argmin(dist))`:** Añade el índice del clúster más cercano (el que tiene la distancia mínima) a `pred`.

```python
def pred_cluster(X, clusters):
    pred = []
    for i in range(X.shape[0]):
        dist = []
        for j in range(k):
            dist.append(distance(X[i], clusters[j]['center']))
        pred.append(np.argmin(dist))
    return pred
```

### Paso 8: Asignar, Actualizar y Predecir los Centros de Clúster

Asignamos puntos a los clústeres, actualizamos los centroides y predecimos las etiquetas de clúster finales.

- **`assign_clusters(X, clusters)`:** Asigna los puntos de datos a los centroides más cercanos.
- **`update_clusters(X, clusters)`:** Recalcula los centroides.
- **`pred_cluster(X, clusters)`:** Predice los clústeres finales para todos los puntos de datos.

```python
clusters = assign_clusters(X, clusters)
clusters = update_clusters(X, clusters)
pred = pred_cluster(X, clusters)
```

### Paso 9: Graficar los Puntos de Datos con los Centros de Clúster Predichos

Finalmente, graficamos los puntos de datos coloreados según sus clústeres predichos, junto con los centroides actualizados.

- **`center = clusters[i]['center']`:** Recupera el centro (centroide) del clúster actual.
- **`plt.scatter(center[0], center[1], marker='^', c='red')`:** Grafica el centro del clúster como un triángulo rojo (marcador `^`).

```python
plt.scatter(X[:, 0], X[:, 1], c=pred)
for i in clusters:
    center = clusters[i]['center']
    plt.scatter(center[0], center[1], marker='^', c='red')
plt.show()
```

**Salida:** Se muestra el resultado final del Clustering K-Means: los puntos de datos están coloreados según el clúster al que pertenecen, y los centroides finales se muestran como triángulos rojos.

## Desafíos del Clustering K-Means

El algoritmo K-Means tiene las siguientes limitaciones:

- **Elegir el número correcto de clústeres** (*k*): Uno de los mayores desafíos es decidir cuántos clústeres usar.
- **Sensible a los centroides iniciales:** Los clústeres finales pueden variar dependiendo de la colocación aleatoria inicial de los centroides.
- **Clústeres no esféricos:** K-Means asume que los clústeres son esféricos y de igual tamaño. Esto puede ser un problema cuando los clústeres reales en los datos tienen formas o densidades diferentes.
- **Valores atípicos (*Outliers*):** K-Means es sensible a los valores atípicos, que pueden distorsionar el centroide y, en última instancia, los clústeres.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

2. Cargar los datos
Generalmente desde un archivo CSV.


```python
#df = pd.read_csv("ventaCSV.csv")
#SE CAMBIA AL SIGUIENTE METODO PARA LEER ARCHIVO EXCEL
df = pd.read_excel("venta2excel.xlsx", engine='openpyxl')

#Ver primeras filas:
print(df.head())
```

```python
#SE TIENE PROBLEMA AL SEPARAR DATOS
#SE PROCEDE A EXTRAER LAS COLUMNAS VERDADERAS

print("df.columns = ")
print (df.columns)
#columnas = str(df.columns)
columnas = df.columns[0]
columnas = columnas.split(',')
print("\ncolumnas = ")
print (columnas)

#AHORA COLUMNAS TIENE LA LISTA DE COLUMNAS DE NUESTRO DF
```

```python
#SE PROCEDE A PARTICIONAR LOS VALORES DE LAS COLUMNAS EN SUBCOLUMNAS SIGUIENDO
#EL SIGUIENTE EJEMPLO
#df[['id', '']] = df['Nombre-Apellido'].str.split('-', expand=True)
df[columnas] = df[df.columns[0]].str.split(',',expand=True)

df
```

```python
#ELIMINAR LA PRIMER COLUMNA, QUE ES LA PROBLEMATICA
df = df.drop(columns=[df.columns[0]])
```

```python
df
```

```python
#Ver últimas filas:
print(df.tail())
```

4. Tipos de variables
Separar variables:

Esto ayuda a decidir cómo analizarlas.


```python
#Numéricas
print(df.select_dtypes(include=np.number))
```

```python
#COMO AHORA NO HAY DATOS NUMERICOS, HAY QUE CONVERTIR LAS COLUMNAS
df[['edad','ventas']] = df[["edad","ventas"]].astype(np.number)

#CONVERSION DE FECHAS
df['fecha_registro'] = pd.to_datetime(df['fecha_registro'], errors='coerce')
```

```python
print(df.select_dtypes(include=np.number))
```

```python
#Categóricas (Almacenan un objeto)
#print(df.select_dtypes(include="object"))
df.select_dtypes(include="object")
```

6. Análisis de valores faltantes
Muy importante.



```python
print(df.isnull().sum())
```

7. Análisis de variables categóricas



```python
#Contar categorías:
print(df["ciudad"].value_counts())

#Aqui es apreciable como hay valores equivalentes,
#pero como la grafia es distinta se ponen como valores diferentes
```

10. Relación entre variables
Variables numéricas



```python
#Matriz de correlación:
corr = df[["edad","ventas"]].corr()

print(corr)
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm")
#Esto muestra qué variables están relacionadas.
```

```python
#Variables categóricas vs numéricas

sns.boxplot(x="nombre", y="edad", data=df)
```

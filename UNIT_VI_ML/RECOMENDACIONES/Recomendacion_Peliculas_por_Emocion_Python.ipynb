Estas se toman como entrada y las películas correspondientes se mostrarán según la emoción. La correspondencia de cada emoción con el género de películas se lista a continuación:

Con base en la emoción ingresada, se seleccionará el género correspondiente y se recomendarán al usuario las **5 mejores películas** de ese género.

## Implementación del Código

El siguiente script realiza scraping de IMDb para obtener los títulos de películas del género correspondiente a la emoción ingresada por el usuario.

> **Nota para Colab:** IMDb puede bloquear solicitudes automatizadas. Si el scraping devuelve resultados vacíos, se activa automáticamente un respaldo con películas de ejemplo para cada género.

````python
import requests
from bs4 import BeautifulSoup
import re

# Diccionario que mapea emociones a URLs de IMDb
URLS = {
    "Drama":   'https://www.imdb.com/search/title/?title_type=feature&genres=drama',
    "Action":  'https://www.imdb.com/search/title/?title_type=feature&genres=action',
    "Comedy":  'https://www.imdb.com/search/title/?title_type=feature&genres=comedy',
    "Horror":  'https://www.imdb.com/search/title/?title_type=feature&genres=horror',
    "Crime":   'https://www.imdb.com/search/title/?title_type=feature&genres=crime',
}

def main(emotion):
    url = URLS.get(emotion)
    print("ok", url)
    if not url:
        print("Emoción no válida.")
        return []

    headers = {
        'User-Agent':
        'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }

    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()  # Verificamos errores HTTP
    except requests.RequestException as e:
        print(f"Error al obtener datos: {e}")
        return []

    soup = BeautifulSoup(response.text, "lxml")

    # Extraemos los títulos de las películas
    titles = [a.get_text() for a in soup.find_all('a', href=re.compile(r'/title/tt\d+/'))]
    return titles


# Función principal (Driver Function)
if __name__ == '__main__':
    emotion = input("Ingresa la emoción (Drama / Action / Comedy / Horror / Crime): ").strip()
    movie_titles = main(emotion)

    if not movie_titles:
        print("No se encontraron títulos.")
    else:
        max_titles = 14 if emotion in ["Drama", "Action", "Comedy", "Horror", "Crime"] else 12
        for title in movie_titles[:max_titles]:
            print(title)
````

### El código anterior ya no funciona. Esta es la versión actualizada a la estructura actual de IMDB.

````python
import requests
from bs4 import BeautifulSoup

# Diccionario que mapea emociones a URLs de IMDb
URLS = {
    "Drama":   'https://www.imdb.com/search/title/?title_type=feature&genres=drama',
    "Action":  'https://www.imdb.com/search/title/?title_type=feature&genres=action',
    "Comedy":  'https://www.imdb.com/search/title/?title_type=feature&genres=comedy',
    "Horror":  'https://www.imdb.com/search/title/?title_type=feature&genres=horror',
    "Crime":   'https://www.imdb.com/search/title/?title_type=feature&genres=crime',
}

def main(emotion):
    url = URLS.get(emotion)
    if not url:
        print("Emoción no válida.")
        return []

    # Importante: Añadir 'Accept-Language' para asegurar la estructura estándar en inglés
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
        'Accept-Language': 'en-US,en;q=0.5'
    }

    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
    except requests.RequestException as e:
        print(f"Error al obtener datos: {e}")
        return []

    soup = BeautifulSoup(response.text, "html.parser")

    # IMDb actual utiliza etiquetas h3 con la clase 'ipc-title__text' para los nombres de películas
    movie_elements = soup.find_all('h3', class_='ipc-title__text')

    titles = []
    for element in movie_elements:
        text = element.get_text().strip()
        # Filtrar textos que correspondan a los títulos (suelen empezar con el número de ranking, ej: "1. The Dark Knight")
        if text and text[0].isdigit() and '.' in text:
            # Limpiamos el número del ranking si solo quieres el nombre de la película
            clean_title = text.split('.', 1)[1].strip()
            titles.append(clean_title)

    return titles


if __name__ == '__main__':
    # Capitalizamos la primera letra para evitar errores si el usuario escribe en minúsculas
    emotion = input("Ingresa la emoción (Drama / Action / Comedy / Horror / Crime): ").strip().capitalize()

    movie_titles = main(emotion)

    if not movie_titles:
        print("No se encontraron títulos. IMDb podría estar protegiendo el sitio o cambió su estructura.")
    else:
        max_titles = 14 if emotion in ["Drama", "Action", "Comedy", "Horror", "Crime"] else 12
        print(f"\n--- Top Películas de {emotion} ---")
        for title in movie_titles[:max_titles]:
            print(title)
````

## Versión Interactiva sin `input()` — Ideal para Colab

En Google Colab la función `input()` puede ser incómoda de usar dentro de scripts. La siguiente celda permite probar el sistema directamente cambiando el valor de la variable `emocion_elegida`. También incluye un respaldo con películas de ejemplo en caso de que IMDb bloquee la solicitud.

````python
import requests
from bs4 import BeautifulSoup
import re

# ── Cambia este valor para probar distintas emociones ─────────────
# Opciones válidas: "Drama", "Action", "Comedy", "Horror", "Crime"
emocion_elegida = "Horror"
# ─────────────────────────────────────────────────────────────────

# Películas de respaldo por género (en caso de bloqueo de IMDb)
RESPALDO = {
    "Drama": [
        "The Shawshank Redemption (1994)",
        "Schindler's List (1993)",
        "Forrest Gump (1994)",
        "The Green Mile (1999)",
        "Good Will Hunting (1997)",
        "A Beautiful Mind (2001)",
        "The Pursuit of Happyness (2006)",
    ],
    "Action": [
        "The Dark Knight (2008)",
        "Mad Max: Fury Road (2015)",
        "Inception (2010)",
        "Gladiator (2000)",
        "Die Hard (1988)",
        "John Wick (2014)",
        "Mission: Impossible – Fallout (2018)",
    ],
    "Comedy": [
        "The Grand Budapest Hotel (2014)",
        "Superbad (2007)",
        "Home Alone (1990)",
        "Ferris Bueller's Day Off (1986)",
        "Groundhog Day (1993)",
        "The Hangover (2009)",
        "Bridesmaids (2011)",
    ],
    "Horror": [
        "The Shining (1980)",
        "Get Out (2017)",
        "Hereditary (2018)",
        "A Quiet Place (2018)",
        "It (2017)",
        "The Conjuring (2013)",
        "Halloween (1978)",
    ],
    "Crime": [
        "The Godfather (1972)",
        "Pulp Fiction (1994)",
        "Goodfellas (1990)",
        "The Departed (2006)",
        "No Country for Old Men (2007)",
        "Fargo (1996)",
        "Heat (1995)",
    ],
}

URLS = {
    "Drama":  'https://www.imdb.com/search/title/?title_type=feature&genres=drama',
    "Action": 'https://www.imdb.com/search/title/?title_type=feature&genres=action',
    "Comedy": 'https://www.imdb.com/search/title/?title_type=feature&genres=comedy',
    "Horror": 'https://www.imdb.com/search/title/?title_type=feature&genres=horror',
    "Crime":  'https://www.imdb.com/search/title/?title_type=feature&genres=crime',
}

def recomendar_peliculas(emotion):
    url = URLS.get(emotion)
    if not url:
        print(f"Emoción '{emotion}' no válida. Opciones: {list(URLS.keys())}")
        return

    print(f"Buscando películas de género: {emotion}")
    print(f"   URL: {url}\n")

    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                      'AppleWebKit/537.36 (KHTML, like Gecko) '
                      'Chrome/91.0.4472.124 Safari/537.36'
    }

    titulos = []
    try:
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, "lxml")
        titulos = [
            a.get_text().strip()
            for a in soup.find_all('a', href=re.compile(r'/title/tt\d+/'))
            if a.get_text().strip()
        ]
        # Filtramos duplicados manteniendo el orden
        vistos = set()
        titulos_unicos = []
        for t in titulos:
            if t not in vistos and len(t) > 1:
                vistos.add(t)
                titulos_unicos.append(t)
        titulos = titulos_unicos
    except requests.RequestException as e:
        print(f"No se pudo acceder a IMDb ({e}). Usando lista de respaldo.")

    # Si el scraping no devolvió resultados útiles, usamos el respaldo
    if not titulos:
        print("IMDb no devolvió títulos (posible bloqueo). Usando lista de respaldo.\n")
        titulos = RESPALDO.get(emotion, [])

    max_titulos = 14 if emotion in URLS else 12
    print(f"Top películas recomendadas — Género: {emotion}")
    print("-" * 45)
    for i, titulo in enumerate(titulos[:max_titulos], start=1):
        print(f"  {i:2d}. {titulo}")


# Ejecutamos la recomendación
recomendar_peliculas(emocion_elegida)
````
